```python
import matplotlib.pyplot as plt
import numpy as np
import os, json
import pandas as pd
import seaborn as sns
import json
```

```python
def json_to_df(path_to_json):
    # Source - https://stackoverflow.com/a
    # Posted by Scott, modified by community. See post 'Timeline' for change history
    # Retrieved 2026-01-26, License - CC BY-SA 3.0
    
    json_files = [pos_json for pos_json in os.listdir(path_to_json) if pos_json.endswith('.json')]
    df_list = []
    for f in json_files:
        with open(path_to_json+f) as json_data:
            data = json.load(json_data)
            json_data.close()
    
        hyperparams = ['D', 'N', 'K', 'n_clients', 'lambda', 'lr', 'seed', 'runtime_sec']
        results = {k: v for k, v in data.items() if k not in hyperparams}

        rows = []
        for key, value in results.items():
            parts = key.split('_vs_')
            algorithm = parts[0]
            metric = parts[1]
        
            # handle augmentation
            if metric.endswith('_aug'):
                algorithm += '_aug'
                metric = metric.replace('_aug', '')
        
            rows.append({
                **{k: data[k] for k in hyperparams},
                'algorithm': algorithm,
                'metric': f'err_{metric}',
                'value': value
            })
        
        df_list.append(pd.DataFrame(rows))
    return pd.concat(df_list, ignore_index=True)
```

```python
def plot_metric_line(
    df,
    x_label,
    metric,
    col,
    algorithms=None,
    # custom_palette=custom_palette,
    label_map=None,
    # y_label_map=y_label_map,
    x_label_map=None,
    y_limits=None,
    sharey=True,
    height=4,
    aspect=1,
    save=False,
    fig_name=None
):
    assert metric in df.columns, f"Metric '{metric}' not found in DataFrame."

    df_plot = df.copy()
    if algorithms is not None:
        df_plot = df_plot[df_plot["algorithm"].isin(algorithms)]

    # Create FacetGrid
    g = sns.FacetGrid(
        df_plot,
        col=col,
        height=height,
        aspect=aspect,
        sharey=sharey
    )

    # Plot with lineplot
    g.map_dataframe(
        sns.lineplot,
        x=x_label,
        y=metric,
        hue="algorithm",
        style="algorithm",
        marker="o",
        markersize=8,
        err_style="bars",
        errorbar=("se", 1),
        err_kws={"linewidth": 2.5},
        linewidth=2,
        # palette=custom_palette
    )

    # Set x-axis ticks to be consistent
    x_ticks = sorted(df_plot[x_label].unique())
    for ax in g.axes.flat:
        ax.set_xticks(x_ticks)
        ax.tick_params(axis='both', which='major', labelsize=20) 
        ax.grid(True, axis="y", linestyle="--", alpha=0.7)
        ax.grid(False, axis="x")

    # Set subplot titles and axis labels
    g.set_titles(f"{col} = {{col_name}}", size=24)
    g.set_axis_labels(x_label, r"$\it{" + metric + "}$", fontsize=20)

    # Use mapped label 
    # y_label = y_label_map.get(metric, r"$\it{" + metric + "}$") if y_label_map else r"$\it{" + metric + "}$"
    # g.set_axis_labels(x_label, y_label, fontsize=24)

    # Set y-limits per subplot if specified
    if y_limits:
        for ax, d_val in zip(g.axes.flat, df_plot[col].unique()):
            if d_val in y_limits:
                ax.set_ylim(*y_limits[d_val])

    # Set x-label only on bottom row
    for ax in g.axes.flat:
        ax.set_xlabel("")
    mid_index = len(g.axes.flat) // 2
    # if x_label_map : x_label = x_label_map[x_label]
    # g.axes.flat[mid_index].set_xlabel(r"$\it{" + x_label + "}$")
    if x_label == "p_out":
        x_label_latex = r"$p_{\mathrm{out}}$"
    elif x_label == "alpha":
         x_label_latex = r"$\alpha$"
    else:
        x_label_latex = r"$\it{" + x_label + "}$"

    g.axes.flat[mid_index].set_xlabel(x_label_latex, fontsize=24)

    # Handle legend
    handles, labels = g.axes.flat[-1].get_legend_handles_labels()
    if label_map:
        labels = [label_map.get(label, label) for label in labels]

    g.fig.legend(
        handles,
        labels,
        loc="lower center",
        ncol=len(set(labels)),
        frameon=False,
        fontsize=21,
        bbox_to_anchor=(0.5, -0.2),
        columnspacing=0.8
    )

    # Adjust layout to fit legend
    g.fig.subplots_adjust(bottom=0.22)

    if save and fig_name:
        plt.savefig(fig_name, bbox_inches="tight", dpi=300)

    plt.show()
```
